```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/creditcard.csv")
```

```python
class_counts = df["Class"].value_counts()

class_counts.index = ["Normal", "Fraud"]

class_counts
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

bars = plt.bar(
    class_counts.index,
    class_counts.values,
    color=["#4CAF50", "#E53935"]
)

# Add values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f"{int(height):,}",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

plt.title("Distribution of Credit Card Transactions",
          fontsize=14,
          fontweight="bold")

plt.xlabel("Transaction Type", fontsize=12)
plt.ylabel("Number of Transactions", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Amount"], bins=50)

plt.title("Distribution of Transaction Amount")

plt.xlabel("Transaction Amount")

plt.ylabel("Frequency")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

plt.show()
```

```python
normal = df[df["Class"] == 0]["Amount"]
fraud = df[df["Class"] == 1]["Amount"]
```

```python
plt.figure(figsize=(8,5))

plt.boxplot([normal, fraud],
            tick_labels=["Normal", "Fraud"])

plt.title("Transaction Amount Comparison")

plt.ylabel("Transaction Amount")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Time"], bins=50)

plt.title("Distribution of Transaction Time")

plt.xlabel("Time")

plt.ylabel("Frequency")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/raw/creditcard.csv")
```

```python
plt.figure(figsize=(14,10))

sns.heatmap(df.corr(),
            cmap="coolwarm",
            center=0)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12,8))

corr = df[["Time", "Amount", "Class"]].corr()

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm",
            center=0)

plt.title("Correlation between Important Features")

plt.tight_layout()

plt.show()
```
